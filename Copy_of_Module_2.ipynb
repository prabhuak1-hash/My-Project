```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np

words = np.array(["Label", "Encoding", "is", "important", "because", "many", "ML", "algorithms"]).reshape(-1, 1)

encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(words)

print("Categories:", encoder.categories_[0])
print(encoded)
```
